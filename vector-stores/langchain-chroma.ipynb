```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
from langchain_core.documents import Document

# Create langchain documents
doc1 = Document(
    page_content="Love Is Embarrassing explores the regret of ignoring red flags and staying in a relationship despite knowing it was falling apart.",
    metadata={"album": "GUTS", "song": "love is embarrassing"}
)

doc2 = Document(
    page_content="Teenage Dream reflects on the fear of growing up and wondering whether your best years are already behind you.",
    metadata={"album": "GUTS", "song": "teenage dream"}
)

doc3 = Document(
    page_content="Pretty Isn't Pretty discusses the pressure of unrealistic beauty standards and the struggle to feel confident in your own skin.",
    metadata={"album": "GUTS", "song": "pretty isn't pretty"}
)

doc4 = Document(
    page_content="Brutal captures the frustration, anxiety, and confusion that often come with navigating teenage life.",
    metadata={"album": "SOUR", "song": "brutal"}
)

doc5 = Document(
    page_content="Good 4 U is an energetic breakup anthem expressing anger and disbelief after seeing an ex move on so easily.",
    metadata={"album": "SOUR", "song": "good 4 u"}
)

doc6 = Document(
    page_content="Jealousy, Jealousy explores the damaging effects of comparing yourself to others on social media and struggling with insecurity.",
    metadata={"album": "SOUR", "song": "jealousy, jealousy"}
)

doc7 = Document(
    page_content="The Cure reflects on searching for emotional healing after heartbreak while realizing that recovery takes patience and time.",
    metadata={"album": "YSPSFAGSIL", "song": "the cure"}
)

doc8 = Document(
    page_content="What's Wrong With Me explores self-doubt, overthinking, and questioning your own worth during emotionally difficult moments.",
    metadata={"album": "YSPSFAGSIL", "song": "what's wrong with me"}
)

doc9 = Document(
    page_content="Expectations examines the disappointment that comes from holding people to standards they fail to meet, leading to heartbreak and personal growth.",
    metadata={"album": "YSPSFAGSIL", "song": "expectations"}
)

doc10 = Document(
    page_content="Cigarette Smoke reflects on the lingering memories of a relationship, capturing the bittersweet feeling of trying to let go while parts of the past still remain.",
    metadata={"album": "YSPSFAGSIL", "song": "cigarette smoke"}
)
```

```python
docs = [doc1, doc2, doc3, doc4, doc5, doc6, doc7, doc8, doc9, doc10]
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")

vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory="chroma_db",
    collection_name="olivia_rodrigo_songs"
)
```

```python
# add documents to the vector store
vector_store.add_documents(docs)
```

```python
# view documents in the vector store
vector_store.get(include = ['embeddings', 'documents', 'metadatas'])
```

```python
# search for similar documents
vector_store.similarity_search(
    query = "What song is about overthinking and self-doubt?",
    k = 2
)
```

```python
# meta data filtering
results = vector_store.similarity_search_with_score(
    query = "Songs about insecurity",
    filter = {'album': 'SOUR'}
)

for doc, score in results:
    print(doc.metadata)
    print(doc.page_content)
    print(score)
    print("-" * 50)
```
